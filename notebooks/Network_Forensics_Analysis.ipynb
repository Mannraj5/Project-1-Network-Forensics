# Network Forensics Analysis Notebook

This notebook demonstrates the portfolio-ready forensic workflow, including data ingestion, preprocessing, visualization, and anomaly detection with NetFlow-style flow data.

```python
from pathlib import Path
import pandas as pd
from IPython.display import display

from netflow_forensics.data_loader import load_flow_data
from netflow_forensics.analysis import run_forensic_analysis
from netflow_forensics.visualization import save_visualizations
from netflow_forensics.ml import run_anomaly_detection
from netflow_forensics.modeling import run_supervised_classification
from netflow_forensics.preprocessing import preprocess

data_path = Path('data/sample_netflow.csv')
df = load_flow_data(data_path)
df.head()
```

## Preprocessing and Feature Engineering
The notebook applies standard NetFlow preprocessing, feature engineering, and label normalization.

```python
df_preprocessed = preprocess(df)
display(df_preprocessed.head())
print('Columns:', df_preprocessed.columns.tolist())
```

## Forensic Summary
Run the main analysis pipeline to produce dataset summaries and suspicious flow lists.

```python
findings = run_forensic_analysis(df)
for section, data in findings.items():
    print(f'--- {section} ---')
    print(data)
```

## Visualizations
Save the core visualizations to the `docs/plots` directory and preview them here.

```python
save_visualizations(df, Path('docs/plots'))
print('Saved plots to docs/plots')
```

## Machine Learning Anomaly Detection
The following section uses an isolation forest to identify anomalous flows from NetFlow telemetry.

```python
df_scored, anomaly_summary = run_anomaly_detection(df)
print('Anomaly summary:', anomaly_summary)
display(df_scored.sort_values('anomaly_score').head(10))
```

## Supervised Classification
This section evaluates a simple supervised model using available NetFlow features and prints classification performance metrics.

```python
classification = run_supervised_classification(df_preprocessed)
print('Classification accuracy:', classification.get('accuracy'))
print('F1 score:', classification.get('f1_score'))
print('Label classes:', classification.get('label_classes'))
print('Confusion matrix:')
print(classification.get('confusion_matrix'))
```
